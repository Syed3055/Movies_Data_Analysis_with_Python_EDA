```python
numeric_col = df.select_dtypes(include=np.number)
```

```python
numeric_col.drop(columns=['MovieID','ReleaseYear'], axis=1,inplace=True)
```

```python
nc
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
for col in nc:
    plt.figure(figsize=(10,5))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```

```python
for col in nc:
    sns.histplot(df[col],bins=50, kde=True)
    plt.title(f'Histogram of {col}')
    plt.show()
```

```python
categorical_col = df.select_dtypes(include='object')
```

```python
categorical_col
```

```python
Categorical_clumns = ['Genre','Country','Director','LeadActor']
```

```python
Categorical_clumns
```

```python
for col in Categorical_clumns:
    plt.figure(figsize=(10,5))
    sns.countplot(y=df[col], order=df[col].value_counts().index)
    plt.title(f'Countplot of {col}')
    plt.show()
```

```python
for col in Categorical_clumns:
    df[col].value_counts().head(20).plot(kind='bar')
    plt.title(f'Top 20 {col} counts')
    plt.show()
```

```python
df.columns
```

```python
sns.scatterplot(df,x='IMDbRating',y='RottenTomatoesScore')
plt.show()
```

```python
df.columns
```

```python
genre_year_stats = df.groupby(["Genre","ReleaseYear"]).agg({
    "Global_BoxOfficeUSD": "mean",
    "BudgetUSD": "mean",
    "IMDbRating": "mean",
    "RottenTomatoesScore": "mean"
}).reset_index()
```

```python
plt.figure(figsize= (14,8))
sns.lineplot(data=genre_year_stats,x='ReleaseYear',y='Global_BoxOfficeUSD',hue='Genre')
plt.title("Average global box office revenue by genre over the years")
plt.show()
```
